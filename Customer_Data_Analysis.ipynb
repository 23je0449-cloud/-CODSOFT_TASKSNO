```python
# Import Libraries

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Display settings
plt.style.use('ggplot')
%matplotlib inline
```

```python
# Load Dataset

df = pd.read_csv("Mall_Customers.csv")

# Display first five rows
df.head()
```

```python
# Dataset Shape

print("Rows and Columns:", df.shape)

# Dataset Information

df.info()

# Statistical Summary

df.describe()
```

```python
# Check Missing Values

print(df.isnull().sum())

# Check Duplicate Rows

print("Duplicate Rows:", df.duplicated().sum())

# Rename Column

df.rename(columns={"Genre":"Gender"}, inplace=True)

# Verify Column Names

df.columns
```

```python
df.head()
```

```python
# Age Distribution

plt.figure(figsize=(8,5))
sns.histplot(df['Age'], bins=15, kde=True, color='skyblue')

plt.title("Age Distribution of Customers")
plt.xlabel("Age")
plt.ylabel("Number of Customers")

plt.show()
```

### 📌 Insight

- The majority of customers are between **20 and 40 years** of age.
- Young adults represent the largest customer segment.
- This age group is likely to be the primary target for marketing campaigns and promotional offers.

```python
# Gender Distribution

gender = df['Gender'].value_counts()

plt.figure(figsize=(6,6))

plt.pie(
    gender,
    labels=gender.index,
    autopct='%1.1f%%',
    startangle=90
)

plt.title("Gender Distribution")

plt.show()
```

### 📌 Insight

- The customer base consists of both male and female customers.
- The distribution is nearly balanced.
- Marketing campaigns can be designed for both genders while considering their purchasing preferences.

```python
# Annual Income Distribution

plt.figure(figsize=(8,5))

sns.histplot(
    data=df,
    x='Annual Income (k$)',
    bins=15,
    kde=True,
    color='orange'
)

plt.title("Annual Income Distribution")
plt.xlabel("Annual Income (k$)")
plt.ylabel("Number of Customers")

plt.show()
```

### 📌 Insight

- Most customers have an annual income between **40k and 80k**.
- Only a small number of customers belong to the very low or very high income groups.
- The business mainly serves middle-income customers.

```python
# Spending Score Distribution

plt.figure(figsize=(8,5))

sns.histplot(
    data=df,
    x='Spending Score (1-100)',
    bins=15,
    kde=True,
    color='green'
)

plt.title("Spending Score Distribution")
plt.xlabel("Spending Score")
plt.ylabel("Number of Customers")

plt.show()
```

### 📌 Insight

- Spending scores are distributed across the full range.
- Customers show different purchasing behaviors.
- Customer segmentation can help identify high-value customers.

```python
# Annual Income vs Spending Score

plt.figure(figsize=(8,6))

sns.scatterplot(
    data=df,
    x='Annual Income (k$)',
    y='Spending Score (1-100)',
    hue='Gender',
    s=80
)

plt.title("Annual Income vs Spending Score")

plt.show()
```

### 📌 Insight

- Customers with similar incomes often have different spending scores.
- Higher income does not always result in higher spending.
- Businesses should consider customer behavior along with income while planning marketing strategies.

```python
# Average Spending Score by Gender

plt.figure(figsize=(6,5))

sns.barplot(
    data=df,
    x='Gender',
    y='Spending Score (1-100)'
)

plt.title("Average Spending Score by Gender")

plt.show()
```

### 📌 Insight

- The average spending score differs between male and female customers.
- This indicates variation in purchasing behavior.
- Gender-based marketing strategies may improve customer engagement.

```python
# Spending Score by Gender

plt.figure(figsize=(8,5))

sns.boxplot(
    data=df,
    x='Gender',
    y='Spending Score (1-100)'
)

plt.title("Spending Score by Gender")

plt.show()
```

### 📌 Insight

- Spending scores vary among customers of both genders.
- Some customers have exceptionally high spending scores, making them potential premium customers.
- Businesses can design loyalty programs to retain these valuable customers.

```python
# Correlation Heatmap

plt.figure(figsize=(8,6))

numeric_df = df.select_dtypes(include=['number'])

sns.heatmap(
    numeric_df.corr(),
    annot=True,
    cmap='coolwarm',
    linewidths=0.5
)

plt.title("Correlation Heatmap")

plt.show()
```

### 📌 Insight

- The heatmap shows the relationship between numerical variables.
- Age, annual income, and spending score have weak correlations.
- Customer purchasing behavior is influenced by multiple factors rather than a single variable.

```python
# Create Age Groups

df['Age Group'] = pd.cut(
    df['Age'],
    bins=[18,25,35,45,55,70],
    labels=['18-25','26-35','36-45','46-55','56-70']
)

df.head()
```

### 📌 Insight

- Customers have been divided into different age groups.
- This helps businesses understand which age groups contribute the most to sales.
- Age-based segmentation enables targeted marketing campaigns.

```python
plt.figure(figsize=(8,5))

sns.barplot(
    data=df,
    x='Age Group',
    y='Spending Score (1-100)'
)

plt.title("Average Spending Score by Age Group")
plt.xlabel("Age Group")
plt.ylabel("Average Spending Score")

plt.show()
```

### 📌 Insight

- Different age groups show different spending patterns.
- The business can identify the most profitable age segment.
- Personalized promotions can be designed for each age group.

```python
high_value = df[
    (df['Annual Income (k$)'] > 70) &
    (df['Spending Score (1-100)'] > 70)
]

high_value
```

### 📌 Insight

- High-income and high-spending customers are considered premium customers.
- These customers contribute significantly to business revenue.
- Loyalty programs and exclusive offers can improve customer retention.

```python
low_value = df[
    (df['Annual Income (k$)'] < 40) &
    (df['Spending Score (1-100)'] < 40)
]

low_value
```

### 📌 Insight

- Low-value customers spend less despite having lower incomes.
- Discounts and promotional offers may encourage higher spending.
- Businesses can use targeted campaigns to increase customer engagement.

```python
top10 = df.sort_values(
    by='Spending Score (1-100)',
    ascending=False
).head(10)

top10
```

### 📌 Insight

- The top 10 customers have the highest spending scores.
- These customers are valuable assets for the business.
- Retaining these customers should be a business priority.
